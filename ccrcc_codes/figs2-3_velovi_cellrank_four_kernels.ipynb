# Eight-sample input preparation and veloVI + CellRank four-kernel analysis

Keep inputs and outputs in the original analysis directories, not main_codes.

0. Optionally rebuild the eight tumor-sample ladata files from loom, QC, and epithelial-atlas inputs.
1. Merge the eight samples and load the shared stemness_score table exported by figs2-2_shared_stemness_scoring.ipynb; do not rescore these cells or import old DPT/CytoTRACE2 scores.
2. Run scVelo preprocessing and veloVI.
3. Recompute DPT with the lowest-stemness Epi_MIOX cell as root.
4. Combine VelocityKernel, ConnectivityKernel, PseudotimeKernel, and CytoTRACEKernel at equal weights, then run GPCCA.
5. Save the four-kernel results and redraw the Epi_JUN lineage gene-trend heatmap.

The shared loom-to-ladata preparation is included below. This notebook contains only the four-kernel downstream analysis.


```python
import warnings
from pathlib import Path

import anndata as ad
import numpy as np
import pandas as pd
import scanpy as sc
import scvelo as scv
import scvi
from scvi.external import VELOVI
import cellrank as cr
import torch

warnings.filterwarnings("ignore")
scv.settings.verbosity = 2
sc.settings.verbosity = 2
scvi.settings.seed = 42
np.random.seed(42)
```

```python
# Keep results in the original analysis directories, not main_codes.
ANALYSIS_DIR = Path("/mnt/disk18t/lr_xcy/riku/kirc_205/velovi")
ADATA_PATH = ANALYSIS_DIR / "adata_for_cellrank2.h5ad"
LADATA_DIR = ANALYSIS_DIR.parent / "scvelo" / "multi_sample_scv"
STEMNESS_SCORES_PATH = ANALYSIS_DIR.parent / "leiden_detailed" / "dpt_root_from_miox" / "stemness_scores_six_states.tsv"
OUTDIR = ANALYSIS_DIR / "cellrank2_analysis_4kernels"
OUTDIR.mkdir(parents=True, exist_ok=True)

# Rebuild the eight sample-level ladata files from loom by default.
REBUILD_LADATA_FROM_LOOM = True

ROOT_SUBTYPE = "Epi_MIOX"
CLUSTER_KEY = "cell_subtype"
ROOT_SCORE_KEY = "stemness_score"
PSEUDOTIME_KEY = "dpt_pseudotime"

N_TOP_GENES = 2000
MIN_SHARED_COUNTS = 20
N_NEIGHBORS = 30
N_PCS = 30

WEIGHTS = {
    "velocity": 0.25,
    "similarity": 0.25,
    "pseudotime": 0.25,
    "cytotrace": 0.25,
}
N_DRIVER_GENES = 20

VELOVI_MAX_EPOCHS = 500
VELOVI_BATCH_SIZE = 256
VELOVI_LR = 1e-2
VELOVI_WEIGHT_DECAY = 1e-2
```

## Optional shared preparation: loom to eight sample-level ladata files

This is the shared input preparation retained from the original multi_sample_scv notebook. It aligns loom barcodes through the merged raw-barcode reference, intersects them with QC and epithelial-atlas cells, and copies the epithelial PCA, integrated PCA, UMAP, and subtype annotations. All epithelial states are retained at this input stage, matching the original saved sample files; the four-kernel analysis performs its six-state restriction below.

No clustering, stemness scoring, velocity fitting, or transition-matrix calculation is performed in this preparation step. The function writes to the original scvelo/multi_sample_scv directory. REBUILD_LADATA_FROM_LOOM defaults to True, so running this notebook requires the raw loom inputs and rebuilds the eight sample-level ladata files, overwriting any same-named files. Set it to False only to reuse existing sample files.


```python
def prepare_eight_sample_ladatas_from_loom(project_dir, output_dir):
    project_dir = Path(project_dir)
    output_dir = Path(output_dir)
    loom_paths = {
        "p11_2": "datasets/11/result/GSM4630029/velocyto/GSM4630029.loom",
        "p12_2": "datasets/12/GSE171306/result/GSM5222645/velocyto/GSM5222645.loom",
        "p14_2t": "datasets/14/GSE156632/results/GSM4735366_2t/velocyto/GSM4735366_2t.loom",
        "p14_3t": "datasets/14/GSE156632/results/GSM4735368_3t/velocyto/GSM4735368_3t.loom",
        "p14_5t": "datasets/14/GSE156632/results/GSM4735372_5t/velocyto/GSM4735372_5t.loom",
        "p14_6t": "datasets/14/GSE156632/results/GSM4735374_6t/velocyto/GSM4735374_6t.loom",
        "p14_7t": "datasets/14/GSE156632/results/GSM4735375_7t/velocyto/GSM4735375_7t.loom",
        "p3_pr4pt": "datasets/3/GSE178481_RAW/result/GSM5392419_pr4pt/velocyto/GSM5392419_pr4pt.loom",
    }
    barcode_path = project_dir / "adata_merge_raw_barcode.h5ad"
    epithelial_path = project_dir / "leiden_detailed" / "adata_epi.h5ad"
    qc_path = project_dir / "adata_qc.h5ad"
    required_paths = [barcode_path, epithelial_path, qc_path]
    required_paths.extend(project_dir / relative for relative in loom_paths.values())
    missing_paths = [str(path) for path in required_paths if not path.is_file()]
    if missing_paths:
        raise FileNotFoundError("Missing loom-preparation inputs:\n" + "\n".join(missing_paths))

    barcode_reference = sc.read_h5ad(barcode_path)
    epithelial_reference = sc.read_h5ad(epithelial_path)
    qc_reference = sc.read_h5ad(qc_path)
    output_dir.mkdir(parents=True, exist_ok=True)
    summary = []

    for sample, relative_path in loom_paths.items():
        sample_barcodes = barcode_reference[barcode_reference.obs["sample"] == sample]
        sample_epi = epithelial_reference[epithelial_reference.obs["sample"] == sample].copy()
        sample_qc = qc_reference[qc_reference.obs["sample"] == sample].copy()
        loom = sc.read_loom(str(project_dir / relative_path))
        loom.obs["original_barcode"] = loom.obs_names.copy()
        loom.obs_names = (
            loom.obs_names.str.replace(r"^[^:]*:", "", regex=True)
            .str.replace(r"x$", "-1", regex=True)
        )
        barcode_mapping = dict(zip(sample_barcodes.obs["raw_barcode"], sample_barcodes.obs_names))
        loom.obs["mapped_index"] = loom.obs_names.to_series().map(barcode_mapping).values
        loom = loom[loom.obs["mapped_index"].notna()].copy()
        loom.obs_names = loom.obs["mapped_index"].tolist()

        keep_qc = sample_epi.obs_names[sample_epi.obs_names.isin(sample_qc.obs_names)]
        keep_loom = sample_epi.obs_names[sample_epi.obs_names.isin(loom.obs_names)]
        merged = scv.utils.merge(sample_qc[keep_qc].copy(), loom[keep_loom].copy())
        sample_epi = sample_epi[merged.obs_names].copy()
        for basis in ["X_pca", "X_pca_inte", "X_umap"]:
            merged.obsm[basis] = sample_epi.obsm[basis].copy()
        merged.obs["cell_subtype"] = sample_epi.obs["cell_subtype"].copy()

        output_path = output_dir / f"ladata_{sample}.h5ad"
        merged.write_h5ad(output_path)
        summary.append({"sample": sample, "cells": merged.n_obs, "output": str(output_path)})
        print("saved:", output_path, "cells:", merged.n_obs)

    return pd.DataFrame(summary).set_index("sample")


if REBUILD_LADATA_FROM_LOOM:
    ladata_preparation_summary = prepare_eight_sample_ladatas_from_loom(
        ANALYSIS_DIR.parent, LADATA_DIR
    )
    print(ladata_preparation_summary)
```

## Merge the eight samples and load the shared stemness scores

The eight-sample merge retains the original sample order, without additional normalization or barcode suffixes. First run main_codes/figs2/figs2-2_shared_stemness_scoring.ipynb to score the original six-state epithelial population and export leiden_detailed/dpt_root_from_miox/stemness_scores_six_states.tsv. This analysis and the Fig. S3 displays both read that exact table by cell barcode.

Do not recompute stemness in the smaller eight-sample population. Old dpt_pseudotime and CytoTRACE2_Score columns are removed because DPT and CellRank CytoTRACE are recomputed downstream. All retained six-state cells must have finite stemness scores; the three excluded states may lack scores.

Save the prepared input in the original velovi directory and continue with the same in-memory object.


```python
if not STEMNESS_SCORES_PATH.is_file():
    raise FileNotFoundError(
        f"Run main_codes/figs2/figs2-2_shared_stemness_scoring.ipynb to export the shared stemness scores first: {STEMNESS_SCORES_PATH}"
    )

sample_keys = [
    "ladata_p11_2", "ladata_p12_2", "ladata_p14_2t", "ladata_p14_3t",
    "ladata_p14_5t", "ladata_p14_6t", "ladata_p14_7t", "ladata_p3_pr4pt",
]
sample_ladatas = [sc.read_h5ad(LADATA_DIR / f"{key}.h5ad") for key in sample_keys]
adata = ad.concat(sample_ladatas, merge="same", label="sample", keys=sample_keys).copy()
del sample_ladatas

if not adata.obs_names.is_unique:
    raise ValueError("Eight-sample barcodes must be unique before stemness mapping.")

# Do not carry forward unused, manually added scores from the old input preparation.
adata.obs.drop(columns=["dpt_pseudotime", "CytoTRACE2_Score"], errors="ignore", inplace=True)
stemness_table = pd.read_csv(
    STEMNESS_SCORES_PATH,
    sep="\t",
    index_col="cell_barcode",
    dtype={"cell_barcode": str},
    float_precision="round_trip",
)
stemness_scores = pd.to_numeric(stemness_table[ROOT_SCORE_KEY], errors="coerce")
if not stemness_scores.index.is_unique:
    raise ValueError("Stemness reference barcodes must be unique.")
adata.obs[ROOT_SCORE_KEY] = stemness_scores.reindex(adata.obs_names)

retained_for_analysis = ~adata.obs[CLUSTER_KEY].isin(["Epi_AQP2", "Epi_CD44_high", "Epi_CA12"])
retained_scores = adata.obs.loc[retained_for_analysis, ROOT_SCORE_KEY]
if not np.isfinite(retained_scores.to_numpy()).all():
    raise ValueError("Every retained six-state cell must have a finite stemness_score.")

adata.write_h5ad(ADATA_PATH)
print("prepared input:", ADATA_PATH)
print("all cells:", adata.n_obs, "retained six-state cells:", int(retained_for_analysis.sum()))
adata
```

```python
adata = adata[~adata.obs['cell_subtype'].isin(['Epi_AQP2','Epi_CD44_high','Epi_CA12'])].copy()
adata
```

## 1. scVelo preprocessing

这里不用 `scv.pp.filter_and_normalize()`，改成当前版本兼容的手动预处理流程。

```python
assert "spliced" in adata.layers and "unspliced" in adata.layers, "adata 必须包含 spliced/unspliced layers"

use_rep = "X_pca_inte" if "X_pca_inte" in adata.obsm else ("X_pca" if "X_pca" in adata.obsm else None)

scv.pp.filter_genes(adata, min_shared_counts=MIN_SHARED_COUNTS)
scv.pp.normalize_per_cell(adata)
sc.pp.log1p(adata)
sc.pp.highly_variable_genes(adata, n_top_genes=N_TOP_GENES)
adata.raw = adata
adata = adata[:, adata.var.highly_variable].copy()
scv.pp.moments(
    adata,
    n_neighbors=N_NEIGHBORS,
    n_pcs=N_PCS,
    use_rep=use_rep,
)

adata
```

## 2. Train veloVI and generate RNA velocity

```python
assert "Ms" in adata.layers and "Mu" in adata.layers, "scVelo moments 后应生成 Ms/Mu"

VELOVI.setup_anndata(adata, spliced_layer="Ms", unspliced_layer="Mu")
vae = VELOVI(adata)
vae.train(
    max_epochs=VELOVI_MAX_EPOCHS,
    lr=VELOVI_LR,
    weight_decay=VELOVI_WEIGHT_DECAY,
    batch_size=VELOVI_BATCH_SIZE,
    early_stopping=True,
)

vae.save(OUTDIR / "velovi_model", overwrite=True, save_anndata=False)
```

```python
latent_time = vae.get_latent_time(n_samples=25)
velocity = vae.get_velocity(n_samples=25, velo_statistic="mean")
rates = vae.get_rates()

latent_time_arr = latent_time.to_numpy() if hasattr(latent_time, "to_numpy") else np.asarray(latent_time)
velocity_arr = velocity.to_numpy() if hasattr(velocity, "to_numpy") else np.asarray(velocity)

scaling = 20 / np.maximum(latent_time_arr.max(axis=0), 1e-8)
adata.layers["velocity"] = velocity_arr / scaling
adata.layers["latent_time_velovi"] = latent_time_arr
adata.var["fit_alpha"] = (rates["alpha"].to_numpy() if hasattr(rates["alpha"], "to_numpy") else np.asarray(rates["alpha"])) / scaling
adata.var["fit_beta"] = (rates["beta"].to_numpy() if hasattr(rates["beta"], "to_numpy") else np.asarray(rates["beta"])) / scaling
adata.var["fit_gamma"] = (rates["gamma"].to_numpy() if hasattr(rates["gamma"], "to_numpy") else np.asarray(rates["gamma"])) / scaling
adata.var["fit_t_"] = torch.nn.functional.softplus(vae.module.switch_time_unconstr).detach().cpu().numpy() * scaling
adata.layers["fit_t"] = latent_time_arr * scaling[np.newaxis, :]
adata.var["fit_scaling"] = 1.0
```

## 3. Recompute DPT pseudotime with your requested root

```python
assert CLUSTER_KEY in adata.obs.columns
assert ROOT_SCORE_KEY in adata.obs.columns

mask = adata.obs[CLUSTER_KEY].astype(str) == ROOT_SUBTYPE
assert mask.sum() > 0, f"没有找到 {CLUSTER_KEY} == {ROOT_SUBTYPE} 的细胞"

root_scores = pd.to_numeric(adata.obs.loc[mask, ROOT_SCORE_KEY], errors="coerce")
root_cell = root_scores.idxmin()
iroot = int(np.where(adata.obs_names == root_cell)[0][0])

adata.uns["iroot"] = iroot
sc.tl.diffmap(adata)
sc.tl.dpt(adata)
adata.obs[PSEUDOTIME_KEY] = adata.obs["dpt_pseudotime"].copy()

print("root cell:", root_cell)
adata.obs[[CLUSTER_KEY, ROOT_SCORE_KEY, PSEUDOTIME_KEY]].loc[[root_cell]].head()
```

## 4. Make sure neighbors/similarity graph is available

```python
if not ("neighbors" in adata.uns and "connectivities" in adata.obsp and "distances" in adata.obsp):
    if use_rep is None and "X_pca" not in adata.obsm:
        sc.pp.pca(adata, n_comps=N_PCS)
        use_rep = "X_pca"
    sc.pp.neighbors(adata, n_neighbors=N_NEIGHBORS, n_pcs=None if use_rep is not None else N_PCS, use_rep=use_rep)

adata
```

## 5. Build the four CellRank kernels

```python
vk = cr.kernels.VelocityKernel(
    adata,
    attr="layers",
    xkey="Ms" if "Ms" in adata.layers else ("spliced" if "spliced" in adata.layers else "X"),
    vkey="velocity",
).compute_transition_matrix()

sk = cr.kernels.ConnectivityKernel(
    adata,
    conn_key="connectivities",
).compute_transition_matrix()

pk = cr.kernels.PseudotimeKernel(
    adata,
    time_key=PSEUDOTIME_KEY,
).compute_transition_matrix(threshold_scheme="hard")

cytotrace_layer = "matrix" if "matrix" in adata.layers else ("spliced" if "spliced" in adata.layers else None)
ctk = cr.kernels.CytoTRACEKernel(adata)
ctk = ctk.compute_cytotrace(
    layer=cytotrace_layer,
    aggregation="mean",
    n_genes=200,
).compute_transition_matrix(threshold_scheme="hard")

vk.write_to_adata(key="T_velovi")
sk.write_to_adata(key="T_similarity")
pk.write_to_adata(key="T_pseudotime")
ctk.write_to_adata(key="T_cytotrace")
```

```python
w = {
    "velocity": float(WEIGHTS["velocity"]),
    "similarity": float(WEIGHTS["similarity"]),
    "pseudotime": float(WEIGHTS["pseudotime"]),
    "cytotrace": float(WEIGHTS["cytotrace"]),}
w_sum = sum(w.values())
w = {k: v / w_sum for k, v in w.items()}

combined_kernel = (
    float(w["velocity"]) * vk +
    float(w["similarity"]) * sk +
    float(w["pseudotime"]) * pk +
    float(w["cytotrace"]) * ctk).compute_transition_matrix()
combined_kernel.write_to_adata(key="T_combined")

combined_kernel
```

## 6. CellRank downstream analysis

```python
g = cr.estimators.GPCCA(combined_kernel)
g.compute_eigendecomposition()
g.compute_macrostates(cluster_key=CLUSTER_KEY)
g.predict_terminal_states(method="stability", n_cells=30)
g.predict_initial_states(n_states=1, n_cells=30,allow_overlap=True)
g.compute_fate_probabilities()
g.compute_lineage_drivers(cluster_key=CLUSTER_KEY)
g.to_adata()

g
```

```python
terminal_states = g.terminal_states.dropna() if getattr(g, "terminal_states", None) is not None else None
initial_states = g.initial_states.dropna() if getattr(g, "initial_states", None) is not None else None

fp = g.fate_probabilities if getattr(g, "fate_probabilities", None) is not None else None
if fp is not None:
    values = fp.X if hasattr(fp, "X") else np.asarray(fp)
    if hasattr(fp, "names"):
        columns = list(fp.names)
    elif hasattr(fp, "lineage_names"):
        columns = list(fp.lineage_names)
    else:
        columns = [f"lineage_{i}" for i in range(values.shape[1])]
    fate_df = pd.DataFrame(values, index=adata.obs_names, columns=columns)
else:
    fate_df = None

print("terminal states")
display(terminal_states.head() if terminal_states is not None else None)
print("initial states")
display(initial_states.head() if initial_states is not None else None)
print("fate probabilities")
display(fate_df.head() if fate_df is not None else None)
```

## 7. Lineage-driver analysis

Compute and export the top driver genes for each terminal lineage.


```python
lineage_names = list(fate_df.columns) if fate_df is not None else []
lineage_driver_tables = {}
top_driver_genes = {}

for lineage in lineage_names:
    drivers = g.compute_lineage_drivers(lineages=lineage, cluster_key=CLUSTER_KEY)
    lineage_driver_tables[lineage] = drivers
    top_driver_genes[lineage] = list(drivers.head(N_DRIVER_GENES).index)
    drivers.to_csv(OUTDIR / f"lineage_drivers_{lineage}.tsv", sep="\t")

top_driver_genes
```

## 8. Save results and processed adata

```python
adata.uns["four_kernel_pipeline"] = {
    "root_cell": root_cell,
    "root_subtype": ROOT_SUBTYPE,
    "root_score_key": ROOT_SCORE_KEY,
    "pseudotime_key": PSEUDOTIME_KEY,
    "weights": WEIGHTS,
    "lineages": lineage_names,
}

if terminal_states is not None:
    terminal_states.to_csv(OUTDIR / "terminal_states.csv", header=True)
if initial_states is not None:
    initial_states.to_csv(OUTDIR / "initial_states.csv", header=True)
if getattr(g, "lineage_drivers", None) is not None:
    g.lineage_drivers.to_csv(OUTDIR / "lineage_drivers.tsv", sep="\t")
if fate_df is not None:
    fate_df.to_csv(OUTDIR / "fate_probabilities.csv")

adata.write_h5ad(OUTDIR / "adata_cellrank_four_kernels.h5ad")
print("saved to:", OUTDIR)
```

# Redraw Epi_JUN CellRank gene heatmap from saved four-kernel h5ad

This notebook reads `cellrank2_analysis_4kernels/adata_cellrank_four_kernels.h5ad` and redraws the Epi_JUN MS heatmap without overwriting the original output.


```python
from pathlib import Path
import warnings

import matplotlib as mpl
import matplotlib.pyplot as plt
import scanpy as sc
import cellrank as cr

warnings.filterwarnings("ignore")

# Keep text editable in vector outputs.
mpl.rcParams["pdf.fonttype"] = 42
mpl.rcParams["ps.fonttype"] = 42
mpl.rcParams["svg.fonttype"] = "none"
mpl.rcParams["font.family"] = "DejaVu Sans"

sc.settings.verbosity = 2
```

```python
BASE_DIR = Path('/mnt/disk18t/lr_xcy/riku/kirc_205/velovi')
H5AD_PATH = BASE_DIR / 'cellrank2_analysis_4kernels' / 'adata_cellrank_four_kernels.h5ad'
REF_EPI_H5AD_PATH = BASE_DIR.parent / 'leiden_detailed' / 'adata_epi.h5ad'
FIG_DIR = BASE_DIR / 'figures' / 'cellrank2_analysis_4kernels'
FIG_DIR.mkdir(parents=True, exist_ok=True)

PSEUDOTIME_KEY = 'dpt_pseudotime'
CLUSTER_KEY = 'cell_subtype'
GAM_N_KNOTS = 6

OUT_PDF = FIG_DIR / 'gene_heatmap_epijun_trend_genes_sorted_ms_and_lineagemarker_ms_text_redraw.pdf'
OUT_SVG = FIG_DIR / 'gene_heatmap_epijun_trend_genes_sorted_ms_and_lineagemarker_ms_text_redraw.svg'

print('input:', H5AD_PATH)
print('output pdf:', OUT_PDF)
print('output svg:', OUT_SVG)
```

```python
adata = sc.read_h5ad(H5AD_PATH)
print(adata)
print('layers:', list(adata.layers.keys()))
print('obsm:', list(adata.obsm.keys()))
print('obs states:', [c for c in adata.obs.columns if 'state' in c or 'macro' in c or 'lineage' in c])


# Match cell_subtype annotation colors to leiden_detailed/adata_epi.h5ad.
adata_epi_ref = sc.read_h5ad(REF_EPI_H5AD_PATH, backed='r')
ref_subtype_colors = dict(
    zip(
        adata_epi_ref.obs['cell_subtype'].cat.categories,
        adata_epi_ref.uns['cell_subtype_colors'],
    )
)
current_subtypes = list(adata.obs['cell_subtype'].cat.categories)
missing_color_subtypes = [s for s in current_subtypes if s not in ref_subtype_colors]
if missing_color_subtypes:
    raise ValueError(f'Missing cell_subtype colors in reference adata_epi: {missing_color_subtypes}')
adata.uns['cell_subtype_colors'] = [ref_subtype_colors[s] for s in current_subtypes]
print('applied cell_subtype color mapping:')
for subtype, color in zip(current_subtypes, adata.uns['cell_subtype_colors']):
    print(subtype, color)
```

```python
epi_jun_20_sorted = [
    'SLC16A12', 'TRPM3', 'ALDOB', 'MIOX', 'GPX3', 'FXYD2', 'ENPP3',
    'HIF1A', 'PTGER3', 'PKHD1', 'NNMT', 'SCGN', 'KLF6',
    'FOS', 'STAT3', 'TJP1', 'VMP1', 'IRF1', 'ATF3', 'HES1', 'CDKN1A', 'MYC',
    'IFI44L', 'ZBTB20', 'PARD3', 'CXCL2', 'CXCL8', 'IFITM1', 'CCL2', 'OSMR', 'GADD45B',
]

missing = [g for g in epi_jun_20_sorted if g not in adata.var_names]
if missing:
    raise ValueError(f'Missing genes in adata.var_names: {missing}')

model = cr.models.GAM(adata, n_knots=GAM_N_KNOTS)
```

```python
# Write exactly to FIG_DIR instead of letting Scanpy prepend another figures/ directory.
sc.settings.figdir = Path('.')

from matplotlib.patches import Patch

heatmap_result = cr.pl.heatmap(
    adata,
    model=model,
    genes=epi_jun_20_sorted,
    lineages='Epi_JUN',
    time_key=PSEUDOTIME_KEY,
    cluster_key=CLUSTER_KEY,
    show_fate_probabilities=True,
    show_all_genes=True,
    data_key='Ms',
    n_jobs=1,
    figsize=(6, 6),
    gene_order=epi_jun_20_sorted,
    return_figure=True,
)

cluster_grid = heatmap_result[0][0]
fig = cluster_grid.fig

legend_handles = [
    Patch(facecolor=color, edgecolor='none', label=subtype)
    for subtype, color in zip(current_subtypes, adata.uns['cell_subtype_colors'])
]
# Place legend outside the heatmap/colorbar area.
fig.legend(
    handles=legend_handles,
    title='cell_subtype',
    loc='upper left',
    bbox_to_anchor=(1.02, 0.98),
    frameon=False,
    fontsize=8,
    title_fontsize=9,
)

fig.savefig(OUT_PDF, bbox_inches='tight')
fig.savefig(OUT_SVG, bbox_inches='tight')
plt.close(fig)
```
